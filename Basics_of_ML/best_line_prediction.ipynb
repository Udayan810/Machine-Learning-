```python
# The data
km = [1, 2, 3, 4, 5, 6]
minutes = [14, 19, 21, 26, 30, 34]

# Step 1: find w and b using order 1 and order 6
w = (34-14) / (6- 1)
b = minutes[0]- w * km[0]
print("w =", w)
print("b =", b)

# Step 2: the model
def predict(km, w, b):
    return w * km + b

# Step 3: predict NEW orders
print("2.5 km ->", predict(2.5, w, b), "minutes")
print("4.5 km ->", predict(4.5, w, b), "minutes")
print("25 km  ->", predict(25, w, b), "minutes")
```
